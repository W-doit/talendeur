```python
#Import special library
from sklearn.preprocessing import MultiLabelBinarizer
```

```python
#convert strings in to lists
skill_tech_col=skill_tech_col.str.split(',')
skill_tech_col=skill_tech_col.apply(lambda lst: [s.strip() for s in lst])
skill_soft_col=skill_soft_col.str.split(',')
skill_soft_col=skill_soft_col.apply(lambda lst: [s.strip() for s in lst])
```

```python
skill_tech_col.head()
```

```python
skill_soft_col.head()
```

```python
#Creating clean lists before using the binarizer
skill_tech_col_clean = skill_tech_col.apply(
    lambda x: [s.strip().lower() for s in x] if isinstance(x, list) else []
)
skill_soft_col_clean = skill_soft_col.apply(
    lambda x: [s.strip().lower() for s in x] if isinstance(x, list) else []
)

```

```python
skill_tech_col_clean.head()
```

```python
skill_soft_col_clean.head()
```

```python
mbinarizer=MultiLabelBinarizer()
```

```python
#Apply binarizer to tech skills
mbinarizer_tech = MultiLabelBinarizer()
tech_matrix = mbinarizer_tech.fit_transform(skill_tech_col_clean)

#Apply binarizer to soft skills
mbinarizer_soft = MultiLabelBinarizer()
soft_matrix = mbinarizer_soft.fit_transform(skill_soft_col_clean)
```

```python
#Convert to a df
tech_skills_bin = pd.DataFrame(tech_matrix, columns=mbinarizer_tech.classes_, index=job_positions.index)
```

```python
#Convert to df
soft_skills_bin=pd.DataFrame(soft_matrix,columns=mbinarizer.classes_,index=job_positions.index)
```

```python
# Convert soft skills matrix to a df
soft_skills_bin = pd.DataFrame(soft_matrix, columns=mbinarizer_soft.classes_)
display(soft_skills_bin.head())
```

```python
jb_full_skills=pd.concat([job_positions,tech_skills_bin,soft_skills_bin],axis=1)
```

```python
jb_full_skills.head()
```

```python
#dropping the original columns
jb_full_skill_clean=jb_full_skills.drop(columns=['technical_skills','soft_skills'])
```

```python
#Compare the columns of the two dataframes to be used in the cosine similarity process candidate_skills_clean_full and jb_full_skill_clean
skill_col_fullcandidate=candidate_skills_clean_full.columns[3:]
skill_col_fullcandidate = [col.lower() for col in skill_col_fullcandidate]
skill_col_fulljob=jb_full_skill_clean.columns[12:]
skill_col_fulljob = [col.lower() for col in skill_col_fulljob]
```

```python
#Create two dataframes that have only the commmon skills
#order the columns
share_columns_ordered=sorted(share_columns)
df_candidate=candidate_skills_clean_full.copy()
df_candidate.columns = df_candidate.columns.str.lower()
df_candidate=df_candidate[share_columns_ordered]
df_job=jb_full_skill_clean[share_columns_ordered]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
cosine_similarity= cosine_similarity(df_candidate, df_job)
```

```python
import numpy as np
```

```python
for job_idx in range(cosine_similarity.shape[1]):
  job_similarities=cosine_similarity[:,job_idx]
  sorted_indices=np.argsort(job_similarities)[::-1]
  top_5_indices=sorted_indices[:5]
```

```python
job_columns = ['job_post', 'company', 'minimum_work_experience', 'max_work_experience', 'education', 'certifications']
candidate_columns = ['candidate_id', 'first_name', 'surname']

for job_idx in range(cosine_similarity.shape[1]):
    job_similarities = cosine_similarity[:, job_idx]
    sorted_indices = np.argsort(job_similarities)[::-1]
    top_5_indices = sorted_indices[:5]

    print(f"\n--- Job details ---")
    job_row = jb_full_skill_clean.iloc[job_idx]
    for col in job_columns:
        if col in jb_full_skill_clean.columns:
            print(f"{col}: {job_row[col]}")

    print(f"\nTOP 5 CANDIDATES:")
    for rank, candidate_idx in enumerate(top_5_indices, 1):
        similarity_score = job_similarities[candidate_idx]
        candidate_row = candidate_skills_clean_full.iloc[candidate_idx]

        print(f"\n{rank}. Candidate {candidate_skills_clean_full.iloc[candidate_idx]['candidate_id']} (Similitud: {similarity_score:.3f})")
        for col in candidate_columns:
            if col in candidate_skills_clean_full.columns:
                print(f"   {col}: {candidate_row[col]}")
```
